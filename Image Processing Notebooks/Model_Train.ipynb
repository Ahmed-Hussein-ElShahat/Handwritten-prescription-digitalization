```python
import torch
from transformers import (
    TrOCRProcessor,
    VisionEncoderDecoderModel,
    AutoTokenizer,
    AutoFeatureExtractor,
    TrainingArguments,
    Trainer
)
from datasets import load_dataset
from PIL import Image

# Load feature extractor and tokenizer
feature_extractor = AutoFeatureExtractor.from_pretrained("google/vit-base-patch16-384")
tokenizer = AutoTokenizer.from_pretrained("xlm-roberta-base")

# Specify model name (for handwritten text)
model_name = "microsoft/trocr-large-handwritten"
model = VisionEncoderDecoderModel.from_pretrained(model_name)

# Setup processor combining feature extractor and tokenizer
processor = TrOCRProcessor(feature_extractor=feature_extractor, tokenizer=tokenizer)
```

```python
def preprocess_function(examples):
    # Process a batch of examples
    pixel_values_batch = []
    labels_batch = []
    for image, text in zip(examples['image'], examples['text']):
        if not isinstance(image, Image.Image):
            image = Image.open(image).convert("RGB")
        pixel_values = feature_extractor(image, return_tensors="pt").pixel_values[0]
        label = tokenizer(text, return_tensors="pt").input_ids[0]
        pixel_values_batch.append(pixel_values)
        labels_batch.append(label)
    return {"pixel_values": pixel_values_batch, "labels": labels_batch}

# Use batched mapping with a small batch size, no multiprocessing, and disable caching
ds = ds.map(preprocess_function, batched=True, batch_size=4, num_proc=1, load_from_cache_file=False)
```

```python
from transformers import PreTrainedTokenizerBase
import torch

class VisionEncoderDecoderDataCollator:
    def __init__(self, tokenizer: PreTrainedTokenizerBase, label_pad_token_id: int = -100):
        self.tokenizer = tokenizer
        self.label_pad_token_id = label_pad_token_id

    def __call__(self, features):
        # Make a copy of features so as not to modify in-place
        new_features = []
        for feature in features:
            feature = feature.copy()
            # Create a temporary "input_ids" key from "labels" for padding
            if "labels" in feature:
                feature["input_ids"] = feature["labels"]
            new_features.append(feature)

        # Pad the tokenized sequences (i.e. the target sequences)
        batch = self.tokenizer.pad(
            new_features,
            padding=True,
            return_tensors="pt",
            # Ensure that padding uses our desired label_pad_token_id
            pad_to_multiple_of=None,
        )
        # Replace padded "input_ids" with our "labels" and remove the temporary key.
        batch["labels"] = batch["input_ids"].clone()
        # Replace padding token id for labels with label_pad_token_id
        batch["labels"][batch["labels"] == self.tokenizer.pad_token_id] = self.label_pad_token_id
        del batch["input_ids"]
        # Note: pixel_values are already tensors, so we leave them as-is.
        return batch
```

```python
import torch
from transformers import (
    TrOCRProcessor,
    VisionEncoderDecoderModel,
    TrainingArguments,
    Trainer
)
from datasets import load_dataset
from PIL import Image

# Load processor and model
model_name = "microsoft/trocr-large-handwritten"
processor = TrOCRProcessor.from_pretrained(model_name)
model = VisionEncoderDecoderModel.from_pretrained(model_name)

# Load dataset with existing splits
ds = load_dataset("Teklia/KHATT_v1.0")

# Final preprocessing with tensor guarantees
def preprocess(examples):
    images = []
    texts = examples["text"]

    # Process images to tensors
    for img in examples["image"]:
        if not isinstance(img, Image.Image):
            img = Image.open(img).convert("RGB")
        pixel_values = processor.feature_extractor(img, return_tensors="pt").pixel_values[0]
        images.append(pixel_values)

    # Process text to tensor labels
    labels = processor.tokenizer(
        texts,
        padding="max_length",
        max_length=64,
        truncation=True,
        return_tensors="pt"
    ).input_ids

    return {
        "pixel_values": images,  # List of tensors
        "labels": labels
    }

# Apply preprocessing ONCE
encoded_ds = ds.map(
    preprocess,
    batched=True,
    batch_size=8,
    remove_columns=ds["train"].column_names
)

# Custom data collator for tensor stacking
class Collator:
    def __call__(self, batch):
        return {
            "pixel_values": torch.stack([item["pixel_values"] for item in batch]),
            "labels": torch.stack([item["labels"][0] for item in batch])  # Remove extra dimension
        }

# Modern training arguments
training_args = TrainingArguments(
    output_dir="./trocr-finetuned",
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    num_train_epochs=150,
    eval_strategy="epoch",
    save_strategy="epoch",
    fp16=True,
    logging_steps=100,
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,
    report_to="none"
)

# Create Trainer with verified components
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=encoded_ds["train"],
    eval_dataset=encoded_ds["validation"],
    data_collator=Collator(),
    tokenizer=processor.feature_extractor
)

# Start training
trainer.train()

# Save model
trainer.save_model("/content/drive/MyDrive/trocr-finetuned/final_model")
```
